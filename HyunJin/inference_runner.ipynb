## 📂 2. 로컬 파일 업로드 및 압축 풀기
폴더를 통째로 업로드하면 수만 개의 파일 전송 병목으로 인해 1시간 이상 걸릴 수 있습니다. **반드시 ZIP 파일로 압축해서 업로드하세요.**

1. 로컬 컴퓨터에서 `my_code_0514from0508` 폴더 자체를 우클릭하여 하나의 ZIP 파일(`my_code_0514from0508.zip`)로 압축합니다.
2. Colab 좌측 파일 아이콘(📁)을 누르고, 만들어진 `my_code_0514from0508.zip` 파일을 빈 공간에 드래그 앤 드롭으로 업로드합니다.
3. 업로드 로딩 동그라미가 완료될 때까지 기다립니다.
4. 완료되면 아래 셀을 실행하여 1초 만에 압축을 풉니다.

```python
import os
import zipfile

ZIP_PATH = '/content/my_code_0514from0508.zip'
EXTRACT_DIR = '/content/'

if os.path.exists(ZIP_PATH):
    print("📦 압축 파일을 해제합니다...")
    with zipfile.ZipFile(ZIP_PATH, 'r') as zip_ref:
        zip_ref.extractall(EXTRACT_DIR)
    print("✅ 압축 해제 완료! 이제 초고속 추론이 가능합니다.")
else:
    print("⚠️ my_code_0514from0508.zip 파일을 찾을 수 없습니다. Colab 좌측 파일 탭에 업로드가 100% 완료되었는지 확인해주세요.")
```
